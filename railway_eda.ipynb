```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("Libraries loaded successfully")
```

## 1. Load Dataset

```python
df = pd.read_csv("train_details.csv")

print("Dataset loaded successfully")
print("Rows and columns:", df.shape)

display(df.head())
```

## 2. Understand Dataset Structure

```python
print("Number of rows and columns:")
print(df.shape)

print("Column names:")
print(df.columns.tolist())

print("Dataset information:")
df.info()
```

```python
print("Numerical summary:")
display(df.describe())

print("All columns summary:")
display(df.describe(include="all"))
```

## 3. Data Quality Checks

```python
print("Missing values in each column:")
display(df.isnull().sum())

print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

## 4. Data Cleaning

```python
before = df.shape

df = df.drop_duplicates()

print("Shape before removing duplicates:", before)
print("Shape after removing duplicates:", df.shape)
```

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().mean() * 100).round(2)
})

display(missing_report)
```

## 5. Exploratory Data Analysis

The following analysis examines numerical columns, categorical columns,
frequencies, averages, minimum values, maximum values, and possible outliers.

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

print("Numerical columns:")
print(numeric_columns)

if len(numeric_columns) > 0:
    display(df[numeric_columns].describe())
else:
    print("No numerical columns found.")
```

```python
categorical_columns = df.select_dtypes(include="object").columns.tolist()

print("Categorical columns:")
print(categorical_columns)

for column in categorical_columns:
    print(f"Top values in {column}:")
    display(df[column].value_counts().head(10))
```
